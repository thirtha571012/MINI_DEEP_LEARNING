```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import sklearn.datasets
from sklearn.model_selection import train_test_split
```

```python
breast_cancer=sklearn.datasets.load_breast_cancer()
```

```python
print(breast_cancer)
```

```python
data_frame=pd.DataFrame(breast_cancer.data,columns=breast_cancer.feature_names)
data_frame.head()
```

```python
data_frame['label']=breast_cancer.target
data_frame.tail()
```

```python
data_frame.shape
```

```python
data_frame.info()
```

```python
data_frame.isnull().sum()
```

```python
data_frame.describe()
```

```python
data_frame['label'].value_counts()
```

```python
X=data_frame.drop(columns='label')
Y=data_frame['label']
print(X)
print(Y)
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=2)
```

STANDARDIZATION OF DATA

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler=StandardScaler()
X_train_std=scaler.fit_transform(X_train)
X_test_std=scaler.transform(X_test)
```

BUILDING THE NEURAL NETWORK

```python
import tensorflow as tf
tf.random.set_seed(3)
from tensorflow import keras
```

```python
model=keras.Sequential([
    keras.layers.Flatten(input_shape=(30,)),
    keras.layers.Dense(20,activation='relu'),
    keras.layers.Dense(2,activation='sigmoid')
])
```

COMPILING NEURAL NETWORK

```python
model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

TRAINING NEURAL NETWORK

```python
trainer=model.fit(X_train_std,Y_train,validation_split=0.1,epochs=10)
```

VISUALIZATION

A)ACCURACY

```python
plt.plot(trainer.history['accuracy'])
plt.plot(trainer.history['val_accuracy'])
plt.title('model accuracy')
plt.ylabel('accuracy')
plt.xlabel('epoch')

plt.legend(['training data', 'validation data'], loc = 'lower right')
```

B)LOSS

```python
plt.plot(trainer.history['loss'])
plt.plot(trainer.history['val_loss'])
plt.title('model loss')
plt.ylabel('loss')
plt.xlabel('epoch')

plt.legend(['training data', 'validation data'], loc = 'upper right')
```

PERFORMANCE ON TEST DATA

```python
loss,accuracy=model.evaluate(X_test_std,Y_test)
print(accuracy)
```

```python
Y_pred = model.predict(X_test_std)
print(Y_pred)
```

CONVERTING PROBABILITIES INTO CLASS LABELS

```python
Y_pred_labels=[np.argmax(i) for i in Y_pred]
print(Y_pred_labels)
```

THE PREDICTIVE SYSTEM

```python
input_data = (11.76,21.6,74.72,427.9,0.08637,0.04966,0.01657,0.01115,0.1495,0.05888,0.4062,1.21,2.635,28.47,0.005857,0.009758,0.01168,0.007445,0.02406,0.001769,12.98,25.72,82.98,516.5,0.1085,0.08615,0.05523,0.03715,0.2433,0.06563)
input_data_as_numpy_array = np.asarray(input_data)
input_data_reshaped = input_data_as_numpy_array.reshape(1,-1)
input_data_std = scaler.transform(input_data_reshaped)

prediction = model.predict(input_data_std)
print("Predicted probabilities:", prediction)

prediction_label = int(np.argmax(prediction))
print("Predicted class:", prediction_label)

if prediction_label == 0:
    print("The tumor is Malignant")
else:
    print("The tumor is Benign")
```
